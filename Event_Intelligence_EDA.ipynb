# About the Dataset

## Major Event Intelligence Analytics Project

This project is based on a large-scale **Event Intelligence Platform** dataset created for practicing real-world data analysis using Python, Pandas, NumPy, Matplotlib, and Seaborn.

The dataset contains ten interconnected relational tables representing customers, event categories, venues, organizers, events, tickets, bookings, payments, attendance, and weather.

The main objective of this project is to simulate a real-world analytics environment where raw and imperfect data must be inspected, cleaned, validated, combined, and converted into meaningful business insights.

## Dataset Structure

The project contains ten relational tables:

* **Customers** - customer demographic and registration information.
* **Event Categories** - event category master data.
* **Venues** - venue capacity, type, location, and parking information.
* **Organizers** - organizer details and experience.
* **Events** - event schedule, venue, organizer, expected attendance, and ticket price.
* **Tickets** - ticket type, seat category, pricing, discount, and status.
* **Bookings** - customer bookings, quantity, channel, and booking status.
* **Payments** - payment transactions, amounts, methods, status, and refunds.
* **Attendance** - event check-in/check-out activity.
* **Weather** - event-day weather conditions.

## Relationships Between Tables

```text
Customers ───────────────→ Bookings ───────────────→ Payments
                              │
                              ↓
                           Tickets ───────────────→ Events
                                                     │
                       ┌─────────────────────────────┼──────────────────────┐
                       ↓                             ↓                      ↓
                Event Categories                 Venues                Organizers
                                                     │
                                                     ↓
                                                 Attendance
                                                     │
                                                     ↓
                                                  Weather
```

## Data Quality Challenges

The raw data intentionally contains real-world data-quality problems such as:

* Missing values
* Duplicate records
* Duplicate business identifiers
* Mixed capitalization and whitespace
* Impossible ages
* Negative prices
* Invalid discounts
* Invalid foreign keys / orphan records
* Date inconsistencies
* Invalid payment/refund values
* Checkout timestamps that break chronology
* Impossible weather measurements
* Outliers
* Inconsistent categorical labels

The purpose is to identify these problems table by table and apply suitable cleaning techniques.

````python
df_weather["weather_date"] = pd.to_datetime(df_weather["weather_date"], errors="coerce", dayfirst=True)
````

````python
df_weather = df_weather.drop_duplicates()

for i in ["temperature_c", "rainfall_mm", "humidity_percent", "wind_speed_kmph"]:
    df_weather[i] = pd.to_numeric(df_weather[i], errors="coerce")

df_weather["temperature_c"] = df_weather["temperature_c"].where(
    df_weather["temperature_c"].between(-50, 60)
)
df_weather["rainfall_mm"] = df_weather["rainfall_mm"].where(
    df_weather["rainfall_mm"] >= 0
)
df_weather["humidity_percent"] = df_weather["humidity_percent"].where(
    df_weather["humidity_percent"].between(0, 100)
)
df_weather["wind_speed_kmph"] = df_weather["wind_speed_kmph"].where(
    df_weather["wind_speed_kmph"].between(0, 200)
)

for i in ["temperature_c", "rainfall_mm", "humidity_percent", "wind_speed_kmph"]:
    df_weather[i] = df_weather[i].fillna(df_weather[i].median())

df_weather["weather_condition"] = (
    df_weather["weather_condition"].astype("string").str.strip().str.title()
)
````

### 2. Which cities have the largest venue capacity?

````python
(
    df_venues.groupby("city")
    .agg(
        total_capacity=("capacity", "sum"),
        average_capacity=("capacity", "mean"),
        venue_count=("venue_id", "nunique")
    )
    .sort_values("total_capacity", ascending=False)
)
````

### 3. What is the average ticket price by seat category?

````python
(
    df_tickets.groupby("seat_category")
    .agg(
        average_original_price=("original_price", "mean"),
        average_final_price=("final_price", "mean"),
        average_discount=("discount_percent", "mean"),
        ticket_count=("ticket_id", "count")
    )
    .sort_values("average_final_price", ascending=False)
)
````

````python
(
    df_payment.groupby("payment_method")
    .agg(
        total_payment=("amount", "sum"),
        average_payment=("amount", "mean"),
        transactions=("payment_id", "count")
    )
    .sort_values("total_payment", ascending=False)
)
````
